```python
#import
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('/content/seasonal_agriculture_performance_dataset.xlsx')
display(df.head())
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
# Check missing values
missing = df.isnull().sum()

print(missing)

# Percentage of missing values
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percent
})

missing_table
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

```python
print("Seasons:")
print(df["Season"].value_counts())

print("\nCrops:")
print(df["Crop"].value_counts())

print("\nStates:")
print(df["State"].value_counts())

print("\nIrrigation Methods:")
print(df["Irrigation_Method"].value_counts())
```

```python
seasonal_summary = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Revenue_INR": "mean",
    "Profit_INR": "mean",
    "Rainfall_mm": "mean",
    "Water_Used_m3": "mean",
    "Disease_Pest_Risk_pct": "mean"
}).round(2)

seasonal_summary
```

```python
season_yield = df.groupby("Season")["Yield_Tonnes_Ha"].mean().sort_values(
    ascending=False
)

print(season_yield)
```

```python
plt.figure(figsize=(8,5))

season_yield.plot(kind="bar")

plt.title("Average Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
season_profit = df.groupby("Season")["Profit_INR"].mean().sort_values(
    ascending=False
)

print(season_profit)
```

```python
plt.figure(figsize=(8,5))

season_profit.plot(kind="bar")

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Rainfall_mm"],
    df["Yield_Tonnes_Ha"],
    alpha=0.5
)

plt.title("Rainfall vs Agricultural Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
correlation = df["Rainfall_mm"].corr(df["Yield_Tonnes_Ha"])

print("Correlation between Rainfall and Yield:", round(correlation, 3))
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Avg_Temperature_C"],
    df["Yield_Tonnes_Ha"],
    alpha=0.5
)

plt.title("Temperature vs Agricultural Yield")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
correlation = df["Avg_Temperature_C"].corr(
    df["Yield_Tonnes_Ha"]
)

print("Correlation:", round(correlation, 3))
```

```python
crop_summary = df.groupby("Crop").agg({
    "Yield_Tonnes_Ha": "mean",
    "Production_Tonnes": "mean",
    "Revenue_INR": "mean",
    "Profit_INR": "mean"
}).round(2)

crop_summary
```

```python
best_crop = crop_summary["Yield_Tonnes_Ha"].idxmax()

print("Best Crop based on average yield:", best_crop)
```

```python
plt.figure(figsize=(10,5))

df.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
state_summary = df.groupby("State").agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean",
    "Production_Tonnes": "mean"
}).round(2)

state_summary
```

```python
best_state = state_summary["Yield_Tonnes_Ha"].idxmax()

print("Highest average yield state:", best_state)
```

```python
irrigation_summary = df.groupby("Irrigation_Method").agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean",
    "Water_Used_m3": "mean"
}).round(2)

irrigation_summary
```

```python
plt.figure(figsize=(9,5))

df.groupby("Irrigation_Method")["Yield_Tonnes_Ha"].mean().sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
water_summary = df.groupby("Season")[
    "Water_Efficiency_t_per_1000m3"
].mean().sort_values(ascending=False)

print(water_summary)
```

```python
plt.figure(figsize=(8,5))

water_summary.plot(kind="bar")

plt.title("Water Efficiency by Season")
plt.xlabel("Season")
plt.ylabel("Water Efficiency")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
risk_summary = df.groupby("Season")[
    "Disease_Pest_Risk_pct"
].mean().sort_values(ascending=False)

print(risk_summary)
```

```python
plt.figure(figsize=(8,5))

risk_summary.plot(kind="bar")

plt.title("Disease and Pest Risk by Season")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["Fertilizer_kg_ha"],
    df["Yield_Tonnes_Ha"],
    alpha=0.5
)

plt.title("Fertilizer Usage vs Yield")
plt.xlabel("Fertilizer (kg/ha)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()
```

```python
corr = df["Fertilizer_kg_ha"].corr(
    df["Yield_Tonnes_Ha"]
)

print("Correlation:", round(corr, 3))
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Profit_INR"], bins=30)

plt.title("Distribution of Agricultural Profit")
plt.xlabel("Profit (INR)")
plt.ylabel("Number of Farms")
plt.tight_layout()
plt.show()
```

```python
most_profitable_season = df.groupby("Season")["Profit_INR"].mean().idxmax()

print("Most profitable season:", most_profitable_season)
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

correlation_matrix["Yield_Tonnes_Ha"].sort_values(
    ascending=False
)
```

```python
plt.figure(figsize=(14,10))

plt.imshow(correlation_matrix, cmap="coolwarm", aspect="auto")

plt.colorbar()

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Matrix of Agricultural Variables")

plt.tight_layout()
plt.show()
```

```python
top_farms = df.sort_values(
    "Profit_INR",
    ascending=False
).head(10)

top_farms[
    [
        "Farm_ID",
        "State",
        "District",
        "Crop",
        "Season",
        "Yield_Tonnes_Ha",
        "Revenue_INR",
        "Profit_INR"
    ]
]
```

```python
season_crop = df.pivot_table(
    values="Yield_Tonnes_Ha",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

season_crop
```

```python
season_crop.plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Crop Yield Comparison Across Seasons")
plt.xlabel("Crop")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_yield_season = df.groupby("Season")[
    "Yield_Tonnes_Ha"
].mean().idxmax()

highest_profit_season = df.groupby("Season")[
    "Profit_INR"
].mean().idxmax()

highest_risk_season = df.groupby("Season")[
    "Disease_Pest_Risk_pct"
].mean().idxmax()

best_crop = df.groupby("Crop")[
    "Yield_Tonnes_Ha"
].mean().idxmax()

best_irrigation = df.groupby("Irrigation_Method")[
    "Yield_Tonnes_Ha"
].mean().idxmax()

print("===== KEY FINDINGS =====")
print("Best season for yield:", best_yield_season)
print("Most profitable season:", highest_profit_season)
print("Highest disease/pest risk season:", highest_risk_season)
print("Highest yielding crop:", best_crop)
print("Best irrigation method by yield:", best_irrigation)
```

## Final Dashboard

```python
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
fig.suptitle('Agricultural Performance Dashboard', fontsize=20)

# Plot 1: Average Agricultural Yield by Season
season_yield.plot(kind="bar", ax=axes[0, 0])
axes[0, 0].set_title("Average Agricultural Yield by Season", fontsize=14)
axes[0, 0].set_xlabel("Season", fontsize=12)
axes[0, 0].set_ylabel("Yield (Tonnes/Ha)", fontsize=12)
axes[0, 0].tick_params(axis='x', rotation=0)

# Plot 2: Average Profit by Season
season_profit.plot(kind="bar", ax=axes[0, 1], color='lightgreen')
axes[0, 1].set_title("Average Profit by Season", fontsize=14)
axes[0, 1].set_xlabel("Season", fontsize=12)
axes[0, 1].set_ylabel("Average Profit (INR)", fontsize=12)
axes[0, 1].tick_params(axis='x', rotation=0)

# Plot 3: Average Yield by Crop
df.groupby("Crop")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False).plot(kind="bar", ax=axes[1, 0], color='skyblue')
axes[1, 0].set_title("Average Yield by Crop", fontsize=14)
axes[1, 0].set_xlabel("Crop", fontsize=12)
axes[1, 0].set_ylabel("Yield (Tonnes/Ha)", fontsize=12)
axes[1, 0].tick_params(axis='x', rotation=45)

# Plot 4: Yield by Irrigation Method
df.groupby("Irrigation_Method")["Yield_Tonnes_Ha"].mean().sort_values(ascending=False).plot(kind="bar", ax=axes[1, 1], color='salmon')
axes[1, 1].set_title("Yield by Irrigation Method", fontsize=14)
axes[1, 1].set_xlabel("Irrigation Method", fontsize=12)
axes[1, 1].set_ylabel("Average Yield (Tonnes/Ha)", fontsize=12)
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
